📌 What is Machine Learning?
</h3>

### Simple Definition

Machine Learning (ML) is a technique that enables a computer to **learn from data** instead of being explicitly programmed with every rule.

> **In simple words:**  
> **More Data → Better Learning → Better Prediction**

---

💡 Real-Life Example
</h3>

After learning the pattern, the child can identify a **new cat**.

Machine Learning works in the same way.

<h3 style="background-color:#E3F2FD;
           color:#0D47A1;

           border-radius:5px;">
📌 What is Linear Regression?
</h3>

### Simple Definition

Linear Regression is a **Supervised Machine Learning algorithm** used to **predict a continuous (numeric) value** by finding the **Best Fit Line** between the input (**X**) and the output (**Y**).

> **In simple words:**  
> It learns the relationship between two variables and predicts future values.

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
💡 Real-Life Example
</h3>

Suppose you want to predict a **house price** based on its **area**.

| Area (sq.ft) | Price (₹ Lakhs) |
|--------------|-----------------|
| 800 | 40 |
| 1000 | 50 |
| 1200 | 60 |
| 1500 | 75 |

After learning this pattern, the model can predict the price of a new house.

---

<h3 style="background-color:#FFF8E1;
           color:#F57F17;
           padding:8px;
           border-radius:5px;">
🔄 How It Works
</h3>

```mermaid
flowchart LR
    A([Training Data])
    B[Train Linear Regression Model]
    C[Learn Best Fit Line]
    D[Predict Output]
    E([Prediction])

    A --> B --> C --> D --> E
```

---
line is used to predict the output for new input values.


<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
🎯 Goal of Linear Regression
</h3>

Find the **Best Fit Line** that passes as close as possible to all the data points so that the prediction error is minimum.

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
🧮 Mathematical Intuition
</h3>

Linear Regression assumes that the relationship between the **Input (X)** and **Output (Y)** can be represented by a **straight line**.

### Equation of Best Fit Line

$$
\boxed{Y = mX + c}
$$

Where:

- **Y** → Predicted Output (Target)
- **X** → Input Feature
- **m** → Slope of the line *(change in Y for every 1-unit increase in X)*
- **c** → Intercept *(value of Y when X = 0)*

### Example

$$
\boxed{Y = 5X + 20}
$$

If,

$$
X = 10
$$

Then,

$$
Y = 5(10) + 20 = 70
$$

The model predicts the output as **70**.

<h3 style="background-color:#E3F2FD;

📌 What is Error?
</h3>

### Simple Definition

Error is the **difference between the Actual value and the Predicted value**.

> **In simple words:**  
> Error tells us **how far our prediction is from the actual value.**

### Equation

$$
\boxed{\text{Error} = \text{Actual} - \text{Predicted}}
$$

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">

</h3>

| Actual Price | Predicted Price | Error |
|:------------:|:---------------:|:-----:|
| ₹80 Lakhs | ₹75 Lakhs | ₹5 Lakhs |

The prediction is **₹5 Lakhs away** from the actual price.

📊 Error Metrics
</h3>

| Metric | Full Form | Uses |
|---------|-----------|------|

| MSE | Mean Squared Error | Average Squared Error |
| RMSE | Root Mean Squared Error | Square Root of MSE |
| MAPE | Mean Absolute Percentage Error | Percentage Error |

---

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
📌 Mean Absolute Error (MAE)
</h3>

### Simple Definition

MAE is the **average of absolute errors**.

It treats every error equally.

### Equation

$$
\boxed{
MAE=\frac{\sum |y_i-\hat y_i|}{n}
}
$$

Where,

- $y_i$ → Actual Value
- $\hat y_i$ → Predicted Value
- $n$ → Number of observations

### Example

| Actual | Predicted | Absolute Error |
|:------:|:---------:|:--------------:|
| 10 | 8 | 2 |
| 20 | 22 | 2 |
| 30 | 28 | 2 |

$$
MAE=\frac{2+2+2}{3}=2
$$

---

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
📌 Mean Squared Error (MSE)
</h3>

### Simple Definition

MSE is the **average of squared errors**.

Large errors receive a much higher penalty.

### Equation

$$
\boxed{
MSE=\frac{\sum (y_i-\hat y_i)^2}{n}
}
$$

### Example

| Actual | Predicted | Squared Error |
|:------:|:---------:|:-------------:|
| 10 | 8 | 4 |
| 20 | 22 | 4 |
| 30 | 28 | 4 |

$$
MSE=\frac{4+4+4}{3}=4
$$

---

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
📌 Root Mean Squared Error (RMSE)
</h3>

### Simple Definition

RMSE is simply the **square root of MSE**.

It brings the error back to the original unit.

### Equation

$$
\boxed{
RMSE=\sqrt{\frac{\sum (y_i-\hat y_i)^2}{n}}
}
$$

### Example

If,

$$
MSE=4
$$

Then,

$$
RMSE=\sqrt{4}=2
$$

---

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
📌 Mean Absolute Percentage Error (MAPE)
</h3>

### Simple Definition

MAPE tells the **average error in percentage**.

### Equation

$$
\boxed{
MAPE=
\frac{100}{n}
\sum
\left|
\frac{y_i-\hat y_i}{y_i}
\right|
}
$$

### Example

| Actual | Predicted | Percentage Error |
|:------:|:---------:|:----------------:|
| 100 | 90 | 10% |
| 200 | 180 | 10% |
| 300 | 270 | 10% |

$$
MAPE=\frac{10+10+10}{3}=10\%
$$

---

<h3 style="background-color:#FCE4EC;
           color:#880E4F;
           padding:8px;
           border-radius:5px;">
📈 Comparison of Error Metrics
</h3>

| Metric | Unit | Sensitive to Outliers | Best Use |
|---------|------|----------------------|----------|
| MAE | Same as Target | ❌ No | General Purpose |
| MSE | Squared Unit | ✅ Yes | Model Training |
| RMSE | Same as Target | ✅ Yes | Regression Evaluation |
| MAPE | Percentage | ❌ Less | Business Reports |

---

<h3 style="background-color:#F3E5F5;
           color:#6A1B9A;
           padding:8px;
           border-radius:5px;">
📉 Error Metric Relationship
</h3>

```mermaid
flowchart LR
    A([Actual & Predicted Values])
    B[Calculate Error]
    C[Absolute Error]
    D[Squared Error]

    C --> E[MAE]
    C --> F[MAPE]

    D --> G[MSE]
    G --> H[RMSE]

    A --> B
    B --> C
    B --> D
```


````python
import numpy as np
import matplotlib.pyplot as plt

# Prediction Error
error = np.linspace(0, 10, 200)

# Error Metrics
mae = error
mse = error**2
rmse = np.sqrt(mse)
mape = (error / 10) * 100   # Assuming Actual Value = 10

plt.figure(figsize=(9,5))

# Plot Metrics
plt.plot(error, mae,label='MAE', linewidth=4, linestyle='--')
plt.plot(error, mse, label='MSE',linewidth=3)
plt.plot(error, rmse, label='RMSE', linewidth=2)
plt.plot(error, mape, label='MAPE (%)',  linewidth=3,linestyle=':')

# Labels
plt.title("Comparison of Error Metrics", fontsize=16)
plt.xlabel("Prediction Error", fontsize=12)
plt.ylabel("Metric Value", fontsize=12)

plt.grid(alpha=0.3)
plt.legend(fontsize=11)

plt.tight_layout()
plt.show()
````

<h3 style="background-color:#E3F2FD;
           color:#0D47A1;
           padding:8px;
           border-radius:5px;">
📌 What is R² Score?
</h3>

### Simple Definition

R² Score is an evaluation metric used in **Linear Regression** to measure **how well the model explains the variation in the data.**

> **In simple words:**  
> It tells **how much of the data is explained by the model.**

- Higher R² → Better Model
- Lower R² → Poor Model

The value of **R²** lies between **0 and 1**.

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
💡 Real-Life Example
</h3>

Suppose you want to predict **house prices** based on **house area**.

If the model explains **90%** of the variation in house prices, then

$$
\boxed{R^2 = 0.90}
$$

This means:

- **90%** of the variation is explained by the model.
- **10%** is due to other factors or random error.

---

<h3 style="background-color:#FFF8E1;
           color:#F57F17;
           padding:8px;
           border-radius:5px;">
🔄 How R² Score Works
</h3>

```mermaid
flowchart LR
    A([Actual Values])
    B[Train Model]
    C[Predict Values]
    D[Calculate SSE]
    E[Calculate SST]
    F[Calculate R2]
    G([Evaluate Model])

    A --> B
    B --> C
    C --> D
    A --> E
    D --> F
    E --> F
    F --> G
```

---

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
🧮 Mathematical Intuition
</h3>

R² compares the **prediction error** made by the model with the **total variation** present in the data.

### Equation

$$
\boxed{
R^2
=
1-
\frac{\sum_{i=1}^{n}(y_i-\hat{y}_i)^2}
{\sum_{i=1}^{n}(y_i-\bar{y})^2}
}
$$

Where,

- **$y_i$** → Actual Value
- **$\hat{y}_i$** → Predicted Value
- **$\bar{y}$** → Mean of Actual Values
- **$n$** → Number of observations

---

### Simplified Formula

$$
\boxed{
R^2 = 1 - \frac{SSE}{SST}
}
$$

Where,

- **SSE (Sum of Squared Errors)** → Prediction error made by the model.
- **SST (Total Sum of Squares)** → Total variation present in the data.

---

📈 Visual Intuition

```mermaid
    D[Reduce SSE]
    E([Higher R2 Score])

    A --> B --> C --> D --> E
```

<h3 style="background-color:#E3F2FD;
           color:#0D47A1;
           padding:8px;
           border-radius:5px;">
📌 What is Gradient Descent?
</h3>

### Simple Definition

Gradient Descent is an **optimization algorithm** used to find the **best values of model parameters (weights)** by **minimizing the error (Cost Function).**

> **In simple words:**  
> Gradient Descent keeps adjusting the model until the prediction error becomes as small as possible.

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
💡 Real-Life Example
</h3>

Imagine you are standing on the **top of a mountain** and want to reach the **lowest point (valley)**.

- You look around.
- Move a small step in the downward direction.
- Again check your surroundings.
- Keep repeating this process.
- Finally, you reach the bottom.

Gradient Descent works in exactly the same way. It keeps moving towards the **minimum error**.

---

<h3 style="background-color:#FFF8E1;
           color:#F57F17;
           padding:8px;
           border-radius:5px;">
🔄 How Gradient Descent Works
</h3>

```mermaid
flowchart LR
    A([Initialize Random Weights])
    B[Predict Output]
    C[Calculate Cost]
    D[Compute Gradient]
    E[Update Weights]
    F{Minimum Cost?}
    G([Model Ready])

    A --> B
    B --> C
    C --> D
    D --> E
    E --> F
    F -- No --> B
    F -- Yes --> G
```

---

<h3 style="background-color:#F3E5F5;
           color:#6A1B9A;
           padding:8px;
           border-radius:5px;">
📈 Visual Intuition
</h3>

**Graph:** Cost Function (Parabola)

- **X-axis:** Weight (Parameter)
- **Y-axis:** Cost (Error)

The algorithm starts from a random point and gradually moves towards the **minimum cost**.

</h3>

The goal of Gradient Descent is to find the **parameter values** that produce the **minimum cost**, resulting in the **best possible predictions**.

📖 Cost Function & Mathematical Intuition
</h2>

---

<h3 style="background-color:#E3F2FD;
           color:#0D47A1;
           padding:8px;
           border-radius:5px;">
📌 What is a Cost Function?
</h3>

### Simple Definition

A **Cost Function** measures **how wrong the model's predictions are.**

> **In simple words:**  
> It tells us **how much error the model is making.**

- Higher Cost → Poor Model ❌
- Lower Cost → Better Model ✅

The main goal of Gradient Descent is to **minimize the Cost Function.**

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
💡 Real-Life Example
</h3>

Suppose you are trying to predict **house prices**.

| Actual Price | Predicted Price |
|--------------|-----------------|
| ₹50 Lakhs | ₹48 Lakhs |
| ₹60 Lakhs | ₹65 Lakhs |
| ₹70 Lakhs | ₹68 Lakhs |

Since the predicted prices are not exactly equal to the actual prices, the model has made some **error**.

The Cost Function calculates the **overall error** of all predictions.

---

<h3 style="background-color:#FFF8E1;
           color:#F57F17;
           padding:8px;
           border-radius:5px;">
🔄 How Cost Function Works
</h3>

```mermaid
flowchart LR
    A([Training Data])
    B[Predict Output]
    C[Compare with Actual]
    D[Calculate Error]
    E([Compute Cost])

    A --> B --> C --> D --> E
```

---

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
🧮 Mathematical Intuition
</h3>

For Linear Regression, the most commonly used Cost Function is the **Mean Squared Error (MSE).**

### Cost Function Equation

$$
\boxed{
J(m,c)=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
}
$$

Where,

- **J(m,c)** → Cost Function
- **$y_i$** → Actual Value
- **$\hat{y}_i$** → Predicted Value
- **n** → Number of observations

The square makes all errors **positive** and gives **more penalty to large errors**.

---

### Example

| Actual | Predicted | Error | Squared Error |
|:------:|:---------:|:-----:|:-------------:|
| 10 | 8 | 2 | 4 |
| 20 | 18 | 2 | 4 |
| 30 | 27 | 3 | 9 |

$$
J=\frac{4+4+9}{3}
=\frac{17}{3}
=5.67
$$

The model's average squared error is **5.67**.

---

<h3 style="background-color:#F3E5F5;
           color:#6A1B9A;
           padding:8px;
           border-radius:5px;">
📈 Cost Function Curve
</h3>

The Cost Function forms a **bowl-shaped (convex) curve**.

- **X-axis** → Model Parameters (Weight)
- **Y-axis** → Cost (Error)

The **lowest point** on the curve represents the **minimum cost**, which is the best solution.

> **Insert the Cost Function graph below.**

🎯 Why Gradient Descent?
</h3>

Gradient Descent starts from a **random point** on the Cost Function curve and keeps moving **downhill** until it reaches the **minimum cost**, where the model performs the best.

📖 Learning Rate & Weight Update Equation
</h2>

---

<h3 style="background-color:#E3F2FD;
           color:#0D47A1;
           padding:8px;
           border-radius:5px;">
📌 What is Learning Rate?
</h3>

### Simple Definition

Learning Rate decides **how much the model should update its weights in each step** while minimizing the Cost Function.

> **In simple words:**  
> It controls the **size of each step** taken towards the minimum cost.

- Small Learning Rate → Small steps 🐢
- Large Learning Rate → Large steps 🐇

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
💡 Real-Life Example
</h3>

Imagine you are climbing **down a mountain**.

- If you take **very small steps**, you will reach the bottom safely but slowly.
- If you take **very large steps**, you may skip the valley or even fall.
- Taking **medium-sized steps** helps you reach the bottom quickly and safely.

Learning Rate works in exactly the same way.

---

<h3 style="background-color:#FFF8E1;
           color:#F57F17;
           padding:8px;
           border-radius:5px;">
🔄 Learning Rate Flow
</h3>

```mermaid
flowchart LR
    A([Initialize Weights])
    B[Calculate Cost]
    C[Compute Gradient]
    D[Choose Learning Rate]
    E[Update Weights]
    F([Repeat Until Cost is Minimum])

    A --> B --> C --> D --> E --> F
```

---

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
🧮 Weight Update Equation
</h3>

Gradient Descent updates the weights after every iteration.

### Weight Update Rule

$$
\boxed{
w_{\text{new}}
=
w_{\text{old}}
-
\eta
\frac{\partial J}{\partial w}
}
$$

Similarly, for the bias,

$$
\boxed{
b_{\text{new}}
=
b_{\text{old}}
-
\eta
\frac{\partial J}{\partial b}
}
$$

Where,

- **$w$** → Weight
- **$b$** → Bias (Intercept)
- **$\eta$ (eta)** → Learning Rate
- **$J$** → Cost Function
- **$\frac{\partial J}{\partial w}$** → Gradient (Slope of the Cost Function)

---

<h3 style="background-color:#E1F5FE;
           color:#01579B;
           padding:8px;
           border-radius:5px;">
📌 Why Minus (-) Sign?
</h3>

The gradient points towards the **steepest increase** in the Cost Function.

To **reduce the cost**, we move in the **opposite direction**, so we subtract the gradient.

```mermaid
flowchart LR
    A([High Cost])
    B[Compute Gradient]
    C[Move Opposite to Gradient]
    D([Lower Cost])

    A --> B --> C --> D
```

---

<h3 style="background-color:#F3E5F5;
           color:#6A1B9A;
           padding:8px;
           border-radius:5px;">
📝 Numerical Example
</h3>

Suppose,

$$
w_{\text{old}} = 10
$$

$$
\eta = 0.1
$$

$$
\frac{\partial J}{\partial w} = 4
$$

Using the update rule,

$$
w_{\text{new}}
=
10-(0.1\times4)
$$

$$
\boxed{
w_{\text{new}}=9.6
}
$$

The weight is updated from **10** to **9.6**, moving one step closer to the minimum cost.

📊 Effect of Learning Rate
</h3>

| Learning Rate | Behaviour |
|--------------|-----------|
| Very Small | Slow Learning, takes many iterations |
| Optimal | Fast and Stable Convergence |
| Very Large | Overshoots the minimum and may never converge |

<h3 style="background-color:#FFF3E0;

</h3>

           font-weight:bold;">
📖 Numerical Example (Gradient Descent)
</h2>

---

<h3 style="background-color:#E3F2FD;
           color:#0D47A1;
           padding:8px;
           border-radius:5px;">
📌 Step-by-Step Iterations
</h3>

Suppose we have:

- Initial Weight:

$$
w = 10
$$

- Learning Rate:

$$
\eta = 0.1
$$

The weight update equation is

$$
\boxed{
w_{new}
=
w_{old}
-
\eta
\frac{\partial J}{\partial w}
}
$$

Assume the gradients obtained during each iteration are shown below.

| Iteration | Weight | Gradient | Updated Weight |
|:---------:|:------:|:--------:|:--------------:|
| 1 | 10.0 | 4.0 | 9.6 |
| 2 | 9.6 | 3.0 | 9.3 |
| 3 | 9.3 | 2.0 | 9.1 |
| 4 | 9.1 | 1.0 | 9.0 |
| 5 | 9.0 | 0.0 | Stop |

---

### Iteration 1

Current Weight

$$
w=10
$$

Gradient

$$
\frac{\partial J}{\partial w}=4
$$

Update

$$
w_{new}
=
10-(0.1\times4)
$$

$$
\boxed{
w_{new}=9.6
}
$$

---

### Iteration 2

Current Weight

$$
w=9.6
$$

Gradient

$$
\frac{\partial J}{\partial w}=3
$$

Update

$$
w_{new}
=
9.6-(0.1\times3)
$$

$$
\boxed{
w_{new}=9.3
}
$$

---

### Iteration 3

Current Weight

$$
w=9.3
$$

Gradient

$$
\frac{\partial J}{\partial w}=2
$$

Update

$$
w_{new}
=
9.3-(0.1\times2)
$$

$$
\boxed{
w_{new}=9.1
}
$$

---

### Iteration 4

Current Weight

$$
w=9.1
$$

Gradient

$$
\frac{\partial J}{\partial w}=1
$$

Update

$$
w_{new}
=
9.1-(0.1\times1)
$$

$$
\boxed{
w_{new}=9.0
}
$$

---

### Final Result

The gradient becomes

$$
\boxed{
\frac{\partial J}{\partial w}=0
}
$$

Since the slope is **zero**, the algorithm has reached the **minimum cost**.

Gradient Descent stops updating the weight.

📌 What are the Types of Gradient Descent?
</h3>

Gradient Descent can update the model using:

1. **Entire Dataset** → Batch Gradient Descent
2. **One Data Point at a Time** → Stochastic Gradient Descent (SGD)
3. **Small Group of Data Points** → Mini-Batch Gradient Descent

---

<h3 style="background-color:#FFF8E1;
           color:#F57F17;
           padding:8px;

```mermaid
    B --> C[Batch Gradient Descent]
    B --> D[Stochastic Gradient Descent]
    B --> E[Mini-Batch Gradient Descent]

    C --> F[Use Entire Dataset]
    D --> G[Use One Data Point]
    E --> H[Use Small Batch]
```

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
📖 1. Batch Gradient Descent
</h3>

### Simple Definition

Batch Gradient Descent uses **the entire training dataset** to calculate the gradient before updating the weights.

> **In simple words:**  
> Read the **whole book** first, then make one decision.

### Example

Suppose you have **10,000 training records**.

Batch GD uses **all 10,000 records** before updating the weights once.

**Advantages**

- Stable learning
- Smooth convergence
- Accurate gradient

**Disadvantages**

- Slow for large datasets
- High memory usage

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
📖 2. Stochastic Gradient Descent (SGD)
</h3>

### Simple Definition

SGD updates the weights **after every single training example**.

> **In simple words:**  
> Learn after reading **one page** instead of the whole book.

### Example

Dataset contains

```
10000 Records
```

Update happens like

```
Record 1 → Update

Record 2 → Update

Record 3 → Update

...
```

**Advantages**

- Very fast
- Less memory
- Suitable for huge datasets

**Disadvantages**

- Noisy updates
- Oscillates near minimum

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
📖 3. Mini-Batch Gradient Descent
</h3>

### Simple Definition

Mini-Batch GD divides the dataset into **small batches** and updates weights after each batch.

> **In simple words:**  
> Learn after reading **one chapter** instead of one page or the entire book.

### Example

Suppose

```
10000 Records
```

Batch Size

```
100
```

Updates happen as

```
100 Records → Update

Next 100 → Update

Next 100 → Update
```

**Advantages**

- Faster than Batch GD
- More stable than SGD
- Most commonly used in Deep Learning

**Disadvantages**

- Need to choose batch size

<h3 style="background-color:#E3F2FD;

📌 What is Feature Scaling?
</h3>

### Simple Definition

Feature Scaling is the process of **bringing all features to a similar scale** so that no feature dominates another during model training.

> **In simple words:**  
> Convert all features into a similar range before training the model.

---

<h3 style="background-color:#E8F5E9;

           border-radius:5px;">
💡 Real-Life Example

Notice,

- Age ranges between **20–40**
- Salary ranges between **30,000–80,000**

Since Salary has much larger values, the model gives it **more importance**, even though Age may also be important.

Feature Scaling makes both features comparable.

<h3 style="background-color:#E3F2FD;
           color:#0D47A1;
           padding:8px;
           border-radius:5px;">
📌 What is Min-Max Scaling?
</h3>

### Simple Definition

Min-Max Scaling transforms data into a **fixed range**, usually between **0 and 1**.

> **In simple words:**  
> The smallest value becomes **0**, the largest value becomes **1**, and all other values lie between **0 and 1**.

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
💡 Real-Life Example
</h3>

Suppose student marks are

| Marks |
|-------:|
| 20 |
| 40 |
| 60 |
| 80 |
| 100 |

After Min-Max Scaling

| Marks | Scaled Value |
|-------:|-------------:|
|20|0.00|
|40|0.25|
|60|0.50|
|80|0.75|
|100|1.00|

---

<h3 style="background-color:#FFF8E1;
           color:#F57F17;
           padding:8px;
           border-radius:5px;">
🔄 How It Works
</h3>

```mermaid
flowchart LR
    A([Original Data])
    B[Find Minimum Value]
    C[Find Maximum Value]
    D[Apply Min-Max Formula]
    E([Scaled Data 0 to 1])

    A --> B --> C --> D --> E
```

---

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
🧮 Formula
</h3>

### Min-Max Scaling Formula

$$
\boxed{
x'=
\frac{x-x_{min}}
{x_{max}-x_{min}}
}
$$

Where

- **x** → Original Value
- **xₘᵢₙ** → Minimum Value
- **xₘₐₓ** → Maximum Value
- **x′** → Scaled Value

---

### Example

Suppose

$$
x=60,\qquad
x_{min}=20,\qquad
x_{max}=100
$$

$$
x'
=
\frac{60-20}{100-20}
=
\frac{40}{80}
=
0.5
$$

Therefore,

$$
\boxed{x'=0.5}
$$

</h3>


↓

After Scaling

```

```

### Disadvantages

📖 Standardization (Z-Score Scaling)
</h2>

---

<h3 style="background-color:#E3F2FD;
           color:#0D47A1;
           padding:8px;
           border-radius:5px;">
📌 What is Standardization?
</h3>

### Simple Definition

Standardization transforms the data so that

- Mean = 0
- Standard Deviation = 1

> **In simple words:**  
> It tells us **how far a value is from the average**.

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
💡 Real-Life Example
</h3>

Suppose students scored

| Marks |
|-------:|
|40|
|50|
|60|
|70|
|80|

Average marks

$$
\mu=60
$$

Student scoring

```
80
```

is above average.

Standardization converts this into a **Z-score**.

---

<h3 style="background-color:#FFF8E1;
           color:#F57F17;
           padding:8px;
           border-radius:5px;">
🔄 How It Works
</h3>

```mermaid
flowchart LR
    A([Original Data])
    B[Calculate Mean]
    C[Calculate Standard Deviation]
    D[Apply Z-Score Formula]
    E([Standardized Data])

    A --> B --> C --> D --> E
```

---

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
🧮 Formula
</h3>

### Z-Score Formula

$$
\boxed{
z=
\frac{x-\mu}{\sigma}
}
$$

Where

- **x** → Original Value
- **μ** → Mean
- **σ** → Standard Deviation
- **z** → Standardized Value

---

### Example

Suppose

$$
x=80
$$

$$
\mu=60
$$

$$
\sigma=10
$$

Then

$$
z
=
\frac{80-60}{10}
=
2
$$

Therefore,

$$
\boxed{z=2}
$$

This means the value is **2 standard deviations above the mean**.

---

<h3 style="background-color:#F3E5F5;
           color:#6A1B9A;
           padding:8px;
           border-radius:5px;">
📈 Visual Intuition
</h3>

```
      Mean
        │
        ▼

-3σ  -2σ  -1σ   0   +1σ  +2σ  +3σ
```

           color:#880E4F;

📌 Advantages & Disadvantages
</h3>

### Advantages

- Works well for normally distributed data
- Handles different scales effectively
- Commonly used in Machine Learning

### Disadvantages

- Not completely robust to extreme outliers
- Does not restrict values to a fixed range

📖 Robust Scaling
</h2>

---

<h3 style="background-color:#E3F2FD;
           color:#0D47A1;
           padding:8px;
           border-radius:5px;">
📌 What is Robust Scaling?
</h3>

### Simple Definition

Robust Scaling is a feature scaling technique that uses the **Median** and **Interquartile Range (IQR)** instead of the **Mean** and **Standard Deviation**.

> **In simple words:**  
> It scales the data while **ignoring the effect of outliers**, making it more reliable for datasets with extreme values.

---

<h3 style="background-color:#E8F5E9;
           color:#1B5E20;
           padding:8px;
           border-radius:5px;">
💡 Real-Life Example
</h3>

Suppose the monthly salaries of employees are:

| Salary (₹) |
|------------:|
| 25,000 |
| 30,000 |
| 35,000 |
| 40,000 |
| 5,00,000 |

Here, **₹5,00,000** is an **outlier**.

- **Min-Max Scaling** gets affected by this outlier.
- **Standardization** also shifts because the mean changes.
- **Robust Scaling** uses the **median**, so the outlier has very little effect.

---

<h3 style="background-color:#FFF8E1;
           color:#F57F17;
           padding:8px;
           border-radius:5px;">
🔄 How It Works
</h3>

```mermaid
flowchart LR
    A([Original Data])
    B[Calculate Median]
    C[Calculate Q1 and Q3]
    D[Compute IQR]
    E[Apply Robust Scaling]
    F([Scaled Data])

    A --> B --> C --> D --> E --> F
```

---

<h3 style="background-color:#E8EAF6;
           color:#283593;
           padding:8px;
           border-radius:5px;">
🧮 Mathematical Intuition
</h3>

Robust Scaling subtracts the **Median** from each value and divides it by the **Interquartile Range (IQR)**.

### Formula

$$
\boxed{
x'
=
\frac{x-\text{Median}}
{Q_3-Q_1}
}
$$

Where,

- **x** → Original Value
- **Median** → Middle value of the dataset
- **Q₁** → First Quartile (25%)
- **Q₃** → Third Quartile (75%)
- **IQR = Q₃ − Q₁**
- **x′** → Scaled Value

---

### Example

Dataset

```
10, 20, 30, 40, 100
```

Median

$$
30
$$

First Quartile

$$
Q_1=20
$$

Third Quartile

$$
Q_3=40
$$

Therefore,

$$
IQR=Q_3-Q_1=40-20=20
$$

Now, scale the value

$$
x=40
$$

$$
x'
=
\frac{40-30}{20}
=
0.5
$$

Therefore,

$$
\boxed{x'=0.5}
$$

---

<h3 style="background-color:#F3E5F5;
           color:#6A1B9A;
           padding:8px;
           border-radius:5px;">
📈 Visual Intuition
</h3>

Without Robust Scaling

```
10   20   30   40                          100
```

↓

Robust Scaling focuses mainly on the **middle 50% (IQR)** and reduces the influence of the outlier.

```
      Q1      Median      Q3
       │         │         │
-------|---------|---------|-------
```

           border-radius:5px;">
📌 Advantages & Disadvantages
</h3>

### Advantages

- Works well with outliers
- Uses Median instead of Mean
- Stable scaling for skewed data
- Better than Min-Max Scaling when extreme values are present

### Disadvantages

- Does not scale data to a fixed range
- Slightly more computationally expensive than Min-Max Scaling

---

<h3 style="background-color:#E1F5FE;
           color:#01579B;
           padding:8px;
           border-radius:5px;">
🎯 When to Use Robust Scaling?
</h3>

✅ Dataset contains many outliers

✅ Skewed data distribution

✅ Financial data

✅ Salary data

✅ House price prediction

❌ Avoid when the data is already normally distributed and has very few outliers.

---

<h3 style="background-color:#FFF3E0;
           color:#E65100;
           padding:8px;
           border-radius:5px;">
📊 Comparison with Other Scaling Methods
</h3>

| Feature | Min-Max | Standardization | Robust Scaling |
|----------|:-------:|:---------------:|:--------------:|
| Uses Mean | ❌ | ✅ | ❌ |
| Uses Median | ❌ | ❌ | ✅ |
| Uses Standard Deviation | ❌ | ✅ | ❌ |
| Uses IQR | ❌ | ❌ | ✅ |
| Handles Outliers | ❌ | ⚠️ Moderate | ✅ Best |
| Output Range | 0 to 1 | Mean = 0, Std = 1 | No Fixed Range |

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

# ---------------------------------
# Sample Data (Contains an Outlier)
# ---------------------------------

data = np.array([[10], [20],  [30],   [40],  [50],  [100]])

# ---------------------------------
# Scaling
# ---------------------------------

minmax = MinMaxScaler()
standard = StandardScaler()
robust = RobustScaler()

data_minmax = minmax.fit_transform(data)
data_standard = standard.fit_transform(data)
data_robust = robust.fit_transform(data)

# ---------------------------------
# Plot
# ---------------------------------

fig, ax = plt.subplots(1, 3, figsize=(18,5))

# ---------------- Min-Max ----------------

ax[0].scatter(range(len(data)), data_minmax, s=80)
ax[0].set_title("Min-Max Scaling")
ax[0].set_xlabel("Sample")
ax[0].set_ylabel("Scaled Value")
ax[0].grid(True)

# ---------------- Standard ----------------

ax[1].scatter(range(len(data)), data_standard, s=80)
ax[1].axhline(0, linestyle='--')
ax[1].set_title("Standardization (Z-Score)")
ax[1].set_xlabel("Sample")
ax[1].set_ylabel("Z-Score")
ax[1].grid(True)

# ---------------- Robust ----------------

ax[2].scatter(range(len(data)), data_robust, s=80)
ax[2].axhline(0, linestyle='--')
ax[2].set_title("Robust Scaling")
ax[2].set_xlabel("Sample")
ax[2].set_ylabel("Scaled Value")
ax[2].grid(True)

plt.tight_layout()
plt.show()
````
